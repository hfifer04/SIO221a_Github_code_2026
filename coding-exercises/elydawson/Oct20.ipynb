```matlab
%% In class function for lecture 6 exercise

function out = orthogdemo_ED(n,m)
    out.time = 0:0.01:2*pi;
    t = out.time;
    out.wave1 = cos(m.*t);
    out.wave2 = sin(n.*t);
    out.product = out.wave1 .* out.wave2;
    out.integral = integral(@(t) cos(m.*t).*sin(n.*t), 0, 2*pi);
end

out = orthogdemo_ED(1,2)
```

```matlab
figure(1); clf;
plot(out.time,out.wave1);
hold on;
plot(out.time,out.wave2);
subtitle(sprintf('Integral: %0.3d', out.integral));
```
